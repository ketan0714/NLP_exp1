##3. Prepare the input paragraph

```python
text = """
Travellers enjoy exploring beautiful places because travelling provides new experiences. 
Many travellers visit mountains, beaches, and historical cities during their holidays. 
They travel with friends, families, or sometimes alone, and they often share their 
travel experiences through photographs and videos. Exploring new destinations helps 
people understand different cultures, meet new people, and create memorable moments.
"""

print(text)
```

##4. Load the spaCy English model.

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)
```

##5. Apply whitespace tokenization.

```python
whitespace_tokens = text.split()

print("Whitespace Tokens:")
print(whitespace_tokens)
```

##6. Apply spaCy word tokenization.

```python
spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)
```

##7. Create a punctuation-filtered spaCy token list.

```python
filtered_tokens = [
    token.text for token in doc
    if not token.is_punct
]

print("Punctuation-Filtered Tokens:")
print(filtered_tokens)
```

##8. Apply Porter stemming.

```python
porter = PorterStemmer()

porter_stems = [
    porter.stem(token.lower())
    for token in filtered_tokens
]

print("Porter Stemming:")
print(porter_stems)
```

##9. Apply Snowball stemming.

```python
snowball = SnowballStemmer("english")

snowball_stems = [
    snowball.stem(token.lower())
    for token in filtered_tokens
]

print("Snowball Stemming:")
print(snowball_stems)
```

##10. Apply spaCy lemmatization.

```python
lemmas = [
    token.lemma_
    for token in doc
    if not token.is_punct and not token.is_space
]

print("spaCy Lemmatization:")
print(lemmas)
```

##11. Create a comparison table for selected words.

```python
porter = PorterStemmer()
snowball = SnowballStemmer("english")

filtered_tokens = []
porter_stems = []
snowball_stems = []
lemmas = []

for token in doc:
    if not token.is_punct and not token.is_space:

        word = token.text.lower()

        filtered_tokens.append(word)
        porter_stems.append(porter.stem(word))
        snowball_stems.append(snowball.stem(word))
        lemmas.append(token.lemma_.lower())

comparison = pd.DataFrame({
    
    "Word": filtered_tokens,
    "Porter Stem": porter_stems,
    "Snowball Stem": snowball_stems,
    "spaCy Lemma": lemmas
})

print(comparison)
```

##12. Measure vocabulary/word variation.

```python
words = [token.lower() for token in filtered_tokens]

total_words = len(words)
unique_words = len(set(words))

print("Total words:", total_words)
print("Unique words:", unique_words)
```

##13. Evaluate semantic coherence qualitatively.

```python

examples = [
    "travelling",
    "experiences",
    "exploring",
    "destinations",
    "cultures"
]

observation_rows = []

for word in examples:

    token = next(
        (
            t for t in doc
            if t.is_alpha and t.text.lower() == word
        ),
        None
    )

    if token is not None:

        observation_rows.append({
            "Original": word,
            "Porter Stem": porter.stem(word),
            "Snowball Stem": snowball.stem(word),
            "spaCy Lemma": token.lemma_.lower()
        })

observation_df = pd.DataFrame(observation_rows)

display(observation_df)
```
